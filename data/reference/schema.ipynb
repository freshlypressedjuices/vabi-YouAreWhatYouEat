<b> Standardise schema to prevent bugs when merging into single dataset </b>

Keys:        iso3 (uppercase alpha-3), year (int)

Grain:       one row per iso3-year (no sex/age/urban splits left in)

Names:       snake_case, unit in the name where useful (sugar_g_day)

Missing:     NaN only. Never 0, -999 or "".

Country names: only in data/reference/countries.csv, never used as a join key

Lag:         <decide in the meeting: 5 or 10 years>

File names:  data/cleaned/<source>_clean.csv

<b>Each person works through this checklist in their own cleaning notebook:</b>

- Reduce to one row per iso3-year. Filter out sex, age, urban/rural, education and any other splits. This is the most common cause of bugs, so do it first.
- Create iso3. If the source has names instead of codes, map them (see the snippet below). Uppercase and strip whitespace.
- Make year an integer. Convert dates or strings with .astype(int).
- Rename columns to snake_case with the unit where useful (sugar_g_day, urban_pop_pct).
- Replace sentinels with NaN. Check for 0 that really means missing, -999, "", "NA", "n/a".
- Keep or drop country_name. Names are only for display. The merge will take names from countries.csv, so cleaned files can drop them.
- Run check_schema(df, "<name>") and paste the one-line output into the group chat.
- Save to data/cleaned/<source>_clean.csv and add a markdown cell listing source, filters and what was dropped.
